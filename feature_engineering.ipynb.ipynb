### Задача 2. Преобразование категориальных переменных

1. Для категориальных переменных `short_model`, `transmission`, `region`, `manufacturer`, `state`, `title_status`, `age_category` распечатайте наименование категории и количество уникальных значений в ней.


```python
def cat_age(year):
    if year >= 2020:
        return 'new'
    elif year >= 2010:
        return 'modern'
    else:
        return 'old'

df['age_category'] = df['year'].apply(cat_age)


columns_to_encode = ['short_model', 'transmission', 'region', 'manufacturer', 'state', 'title_status', 'age_category']

for col in columns_to_encode:
    if col in df.columns:
        print(f"Категория: {col}")
        print(f"Количество уникальных: {df[col].nunique()}")
        print("-" * 30)
```
